#**TASK 15**

```python

import time

max_requests = 3
request_count = 0

prompts = [
    "Hello AI",
    "Explain Python",
    "Explain machine learning",
    "This request will fail"
]

for prompt in prompts:

    request_count += 1

    print("\nRequest:", request_count)

    # Rate limiting
    if request_count > max_requests:

        print("Rate Limit Exceeded!")
        break

    start_time = time.time()

    # Primary model
    if "fail" in prompt.lower():

        response = None

    else:

        response = "Primary Model: " + prompt

    # Fallback model
    if response is None:

        print("Primary Model Failed")
        print("Using Fallback Model")

        response = "Fallback Model: " + prompt

    end_time = time.time()

    print("Response:")
    print(response)

    print(
        "Response Time:",
        round(end_time - start_time, 4),
        "seconds"
    )
```
