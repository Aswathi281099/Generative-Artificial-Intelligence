```python
import torch

# Select GPU if available
if torch.cuda.is_available():

    device = "cuda"

    print("GPU Available")

else:

    device = "cpu"

    print("GPU Not Available")
    print("Using CPU")

# Input
x = torch.randn(4, 10).to(device)

# Split input
a, b = torch.chunk(x, 2, dim=1)

# SwiGLU
output = a * torch.nn.functional.silu(b)

print("\nInput Shape:")
print(x.shape)

print("\nOutput Shape:")
print(output.shape)

print("\nDevice:")
print(device)
```
